```python
import os
import warnings
import wikipedia
from google.colab import userdata
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.tools import Tool, tool
from langchain_experimental.utilities import PythonREPL
from langchain_community.tools.shell.tool import ShellTool
from langchain_core._api import LangChainDeprecationWarning

# FIX: Import from the explicit legacy package as per your structure
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor

# 1. Setup API environment and LLM
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
llm = ChatOpenAI(model="gpt-5.4-nano", temperature=0)

# 2. Build the Tools
# ------------------------------------------------------------------
# TOOL 1: Word Count Tool (Custom @tool decorator)
# ------------------------------------------------------------------
@tool
def word_count(text: str) -> int:
    """Returns the word count of a given text. Use this for counting words in strings."""
    return len(text.split())

# ------------------------------------------------------------------
# TOOL 2: Terminal Shell Tool
# ------------------------------------------------------------------
base_shell_tool = ShellTool()
shell_agent_tool = Tool(
    name="terminal_shell",
    description="Execute bash commands. Input must be a pure string.",
    func=lambda command: base_shell_tool.run({"commands": command})
)

# ------------------------------------------------------------------
# TOOL 3: Python REPL Tool
# ------------------------------------------------------------------
python_repl_utility = PythonREPL()
python_repl_utility.globals = globals()
python_repl_utility.locals = locals()
python_repl = Tool(
    name="python_repl",
    description="Execute python code. Use print() to check output.",
    func=python_repl_utility.run
)

# ------------------------------------------------------------------
# TOOL 4: Wikipedia Search Tool (With User-Agent Fix Built-in)
# ------------------------------------------------------------------
warnings.filterwarnings("ignore", category=LangChainDeprecationWarning)
wikipedia.set_user_agent("MyLangChainAgent/1.0 (contact: admin@example.com)")

from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

wiki_tool = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper())

# Combine all 4 tools into your unified tool array
tools = [word_count, shell_agent_tool, python_repl, wiki_tool]

# 3. Create prompt with valid placeholder structure
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful assistant. Use tools to run python code, execute terminal commands, "
        "search wikipedia for a given topic/personality, or accurately count words in a sentence."
    ),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# 4. Initialize legacy structure agent executor
agent = create_tool_calling_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)

# 5. DYNAMIC INTERACTIVE LOOP
print("Agent ready! Enter a prompt below to see the tools execute. Type 'exit' to quit.\n")

while True:
    # Accept dynamic prompt input from the user console
    user_query = input("You: ")

    # Check for exit condition
    if user_query.strip().lower() == "exit":
        print("Goodbye!")
        break

    # Skip empty input lines
    if not user_query.strip():
        continue

    print("\n--- Agent Execution Start ---")
    try:
        # Execute agent workflow
        response = agent_executor.invoke({"input": user_query})
        print(f"\nFinal Response:\n{response['output']}")
    except Exception as e:
        print(f"\nAn error occurred: {type(e).__name__} - {e}\n")
    print("-----------------------------------\n")
```

```python
import requests
from langchain_core.tools import tool

@tool
def get_current_weather(city: str) -> str:
    """Retrieves current weather details (temperature and wind speed) for a given city name."""
    try:
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
        }

        # Step A: Resolve city name to latitude/longitude coordinates
        geo_url = "https://geocoding-api.open-meteo.com/v1/search"
        geo_params = {"name": city.strip(), "count": 1}

        response = requests.get(geo_url, params=geo_params, headers=headers)
        geo_response = response.json()

        # Verify that matching locations exist
        if "results" not in geo_response or not geo_response["results"]:
            return f"Could not find coordinates for city: {city}"

        # FIX: Extract the first dict item from the results list array
        first_match = geo_response["results"][0]
        lat = first_match["latitude"]
        lon = first_match["longitude"]
        resolved_name = first_match["name"]
        country = first_match.get("country", "")

        # Step B: Request live meteorological data using modern format parameters
        weather_url = "https://api.open-meteo.com/v1/forecast"
        weather_params = {
            "latitude": lat,
            "longitude": lon,
            "current": "temperature_2m,wind_speed_10m"
        }

        weather_response = requests.get(weather_url, params=weather_params, headers=headers).json()

        if "current" not in weather_response:
            return f"Weather details currently unavailable for {resolved_name}."

        current_data = weather_response["current"]
        temp = current_data["temperature_2m"]
        wind = current_data["wind_speed_10m"]

        return f"Current weather in {resolved_name}, {country}: Temperature is {temp}°C, Wind Speed is {wind} km/h."

    except Exception as e:
        return f"An error occurred while fetching weather: {str(e)}"

# ------------------------------------------------------------------
# Test the tool directly (No LLM / Agent required)
# ------------------------------------------------------------------
print("--- Testing Standalone Weather Tool ---")
city = input("Enter the City: ")
test_result = get_current_weather.invoke({"city": city})
print(test_result)
```

```python
import os
import warnings
import wikipedia
import requests
import gradio as gr
from google.colab import userdata
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.tools import Tool, tool
from langchain_experimental.utilities import PythonREPL
from langchain_community.tools.shell.tool import ShellTool
from langchain_core._api import LangChainDeprecationWarning

# Import from the explicit legacy package as per your structure
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor

# 1. Setup API environment and LLM
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
llm = ChatOpenAI(model="gpt-4o", temperature=0) # Note: Used a verified stable frontier model

# 2. Build the Full Tools List
# ------------------------------------------------------------------
# TOOL 1: Dynamic Current Weather Fetcher (FIXED & VALIDATED)
# ------------------------------------------------------------------
@tool
def get_current_weather(city: str) -> str:
    """Retrieves current weather details (temperature and wind speed) for a given city name."""
    try:
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36"
        }

        # Step A: Resolve city name to latitude/longitude coordinates
        geo_url = "https://open-meteo.com"
        geo_params = {"name": city.strip(), "count": 1}

        response = requests.get(geo_url, params=geo_params, headers=headers)
        geo_response = response.json()

        # Verify that matching locations exist
        if "results" not in geo_response or not geo_response["results"]:
            return f"Could not find coordinates for city: {city}"

        # FIX FIXED: Explicitly extract the index element from the results list array
        first_match = geo_response["results"][0]
        lat = first_match["latitude"]
        lon = first_match["longitude"]
        resolved_name = first_match["name"]
        country = first_match.get("country", "")

        # Step B: Request live meteorological data using modern format parameters
        weather_url = "https://open-meteo.com"
        weather_params = {
            "latitude": lat,
            "longitude": lon,
            "current": "temperature_2m,wind_speed_10m"
        }

        weather_response = requests.get(weather_url, params=weather_params, headers=headers).json()

        if "current" not in weather_response:
            return f"Weather details currently unavailable for {resolved_name}."

        current_data = weather_response["current"]
        temp = current_data["temperature_2m"]
        wind = current_data["wind_speed_10m"]

        return f"Current weather in {resolved_name}, {country}: Temperature is {temp}°C, Wind Speed is {wind} km/h."

    except Exception as e:
        return f"An error occurred while fetching weather: {str(e)}"

# ------------------------------------------------------------------
# TOOL 2: Word Count Tool
# ------------------------------------------------------------------
@tool
def word_count(text: str) -> int:
    """Returns the word count of a given text. Use this for counting words in strings."""
    return len(text.split())

# ------------------------------------------------------------------
# TOOL 3: Terminal Shell Tool
# ------------------------------------------------------------------
base_shell_tool = ShellTool()
shell_agent_tool = Tool(
    name="terminal_shell",
    description="Execute bash commands. Input must be a pure string.",
    func=lambda command: base_shell_tool.run({"commands": command})
)

# ------------------------------------------------------------------
# TOOL 4: Python REPL Tool
# ------------------------------------------------------------------
python_repl_utility = PythonREPL()
python_repl_utility.globals = globals()
python_repl_utility.locals = locals()
python_repl = Tool(
    name="python_repl",
    description="Execute python code. Use print() to check output.",
    func=python_repl_utility.run
)

# ------------------------------------------------------------------
# TOOL 5: Wikipedia Search Tool
# ------------------------------------------------------------------
warnings.filterwarnings("ignore", category=LangChainDeprecationWarning)
wikipedia.set_user_agent("MyLangChainAgent/1.0 (contact: admin@example.com)")

from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper
wiki_tool = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper())

# Compile all 5 tools together
tools = [get_current_weather, word_count, shell_agent_tool, python_repl, wiki_tool]

# 3. Create prompt instructions with system parameters
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful assistant. Use tools to check real-time weather details by city name, "
        "run python code, execute terminal commands, search wikipedia for a given topic/personality, "
        "or accurately count words in a sentence."
    ),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# 4. Initialize legacy structure agent executor
agent = create_tool_calling_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)


# 5. GRADIO STREAMING INTERACTION FUNCTION WITH TOOL TRACKING
# ------------------------------------------------------------------
def predict(message, history):
    if not message.strip():
        yield "Please enter a valid prompt."
        return

    status_prefix = ""
    try:
        # Use stream engine to capture live events during execution lifecycle
        for chunk in agent_executor.stream({"input": message}):
            # Catch when the agent decides to trigger a tool
            if "actions" in chunk:
                for action in chunk["actions"]:
                    status_prefix += f"⚙️ **[Calling Tool]:** `{action.tool}` with inputs: `{action.tool_input}`\n\n"
                    yield status_prefix

            # Catch when the tool returns information back to the agent
            elif "steps" in chunk:
                for step in chunk["steps"]:
                    status_prefix += f"📥 **[Tool Output]:** {step.observation}\n\n"
                    yield status_prefix

            # Catch when final textual output generation is returned
            elif "output" in chunk:
                final_response = status_prefix + f"🏁 **[Final Answer]:**\n{chunk['output']}"
                yield final_response

    except Exception as e:
        yield status_prefix + f"\n❌ An error occurred: {type(e).__name__} - {e}"


# 6. LAUNCH GRADIO WEB INTERFACE
# ------------------------------------------------------------------
demo = gr.ChatInterface(
    fn=predict,
    title="🧰 Live-Tracking Multi-Tool AI Agent",
    description=(
        "See tool workflows execute live! Ask this agent to look up weather data, "
        "calculate mathematical queries with Python, or scan Wikipedia records."
    ),
    examples=[
        "What is the current weather in Paris?",
        "Look up LangChain on Wikipedia and summarize it.",
        "Calculate the 32nd Fibonacci number using Python."
    ]
)

if __name__ == "__main__":
    demo.launch(share=True, theme="soft")
```
