```python
# Job-1 => Taks-1 Country Name | Task-2 (Model) | Task-3 (City name)
# Job-2 -> Task-3 (City Name) | Task-4 (Model) | Task-5 -> Final Output (Popular places to visit)
```

```python
from langchain_core.messages import HumanMessage, SystemMessage, AIMessage
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
# Example - 3 -> Multiple Tasks / Chains to run
```

```python
# Define Prompt - Task-1

capital_prompt = ChatPromptTemplate.from_messages(
    [
        (SystemMessage, "You are a geography expert, you know the capitals of all the countries, states etc. You respond with only the name of capital city."),
        (HumanMessage, "What is the capital of {country}?")
    ]
)
country = input("Enter the country name: ")

capital_name =  StrOutputParser()

capital_chain = capital_prompt | model  | capital_name

# Define Prompt - Task-2

places_prompt = ChatPromptTemplate.from_messages(
    [
        (SystemMessage, "You are a geography expert, you know the capitals of all the countries, states etc. You respond with popular places to visit there."),
        (HumanMessage, "List 2 popular places to visit in {capital_name}?")
        ]
)

places_names =  StrOutputParser()

places_chain = places_prompt | model | places_names

print("\nThe ouput from capital chain is : \n", capital_chain.invoke(country))

response = capital_chain.invoke(country)

print("\nThe ouput from places chain is : \n", places_chain.invoke(response))

capital_chain.get_graph().print_ascii()
places_chain.get_graph().print_ascii()
```

```python
country = input("Enter the country name: ")

information_chain = capital_chain | places_chain

response = information_chain.invoke(country)

capital_response = capital_chain.invoke(country)

print("The capital of given country is :",capital_response)

print("The 2 most popular places to visit there are : \n", response)
```

```python
print(f"The graphical representation of the sequential chain named {information_chain} is below\n")

information_chain.get_graph().print_ascii()
```

```python
# Example - Dynamic Chat bot using - Chains (sequential chains)
```

```python
print("--- Country Information Chatbot (Type 'exit' to quit) ---")

while True:
    # 1. Take user input
    country = input("\nEnter the country name: ").strip()

    # 2. Check for the exit condition (case-insensitive)
    if country.lower() == 'exit':
        print("Goodbye! Have a great trip planning session!")
        break

    # 3. Skip empty inputs if the user just presses Enter
    if not country:
        continue

    try:
        # 4. Run the capital chain first to get the capital name
        capital_response = capital_chain.invoke(country)

        # 5. Pass the capital into the places chain
        # (This is more efficient than running the combined chain from scratch)
        places_response = places_chain.invoke(capital_response)

        # 6. Print the formatted results
        print(f"The capital of given country is: {capital_response}")
        print(f"The 2 most popular places to visit there are:\n{places_response}")

    except Exception as e:
        print(f"An error occurred while fetching details: {e}")
```
