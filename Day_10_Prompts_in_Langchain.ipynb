```python
from langchain_core.prompts import PromptTemplate, ChatPromptTemplate
```

```python
# Example - 1 - Prompt Template
```

```python
prompt = "What is the capital of {country}"

template = PromptTemplate.from_template(prompt)

print(type(prompt))
print(type(template))

formatted_input_prompt = template.format(country = "India")
print("\nThe human input prompt is : \n")
print(formatted_input_prompt)

response = model.invoke(formatted_input_prompt)
print("\nThe AI / LLM response is : \n")
print(response.content)
```

```python
from langchain_core.prompts import ChatMessagePromptTemplate
```

```python
# Example - Language Translator
```

```python
chat_prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are a helpful assistant that translates english to hindi, tamil, french etc."),
        ("human", "{text}"),
    ]
)

text = input("Enter the topic: \n")

formatted_input_prompt = chat_prompt_template.format(text = text)

print("\nThe human input prompt is : \n")
print(formatted_input_prompt)

model = ChatOpenAI(model = 'gpt-5.4-nano',api_key=openai_api_key,temperature=0.1)
response = model.invoke(formatted_input_prompt)
print("\nThe AI / LLM response is : \n")
print(response.content)
```
