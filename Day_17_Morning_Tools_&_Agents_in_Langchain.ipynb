```python
from langchain_core.tools import Tool
from langchain_experimental.utilities import PythonREPL

# 1. Instantiate the REPL utility
python_repl = PythonREPL()

# 2. Prevent NameError scopes by injecting the Colab environment globals/locals
python_repl.globals = globals()
python_repl.locals = locals()

# 3. Wrap it inside a LangChain Tool definition
repl_tool = Tool(
    name="python_repl",
    description="A Python shell. Use this to execute user inputs python commands.",
    func=python_repl.run,
)

# 4. Create a reusable function to handle dynamic prompts
def run_dynamic_python(python_code_prompt: str) -> str:
    """
    Accepts a dynamic string containing python code,
    executes it via the LangChain REPL tool, and returns the string output.
    """
    print("--- Executing Dynamic Python Snippet ---")

    # Strip leading/trailing whitespaces from the dynamic input string
    cleaned_code = python_code_prompt.strip()

    # Execute through the tool wrapper
    output = repl_tool.run(cleaned_code)

    return output

# ==========================================
# TEST CASE 1: Recursive Function
# ==========================================
code_fibonacci = """
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n-1) + fibonacci(n-2)

print("Fibonacci result:", fibonacci(10))
"""
result_1 = run_dynamic_python(code_fibonacci)
print(result_1)


# ==========================================
# TEST CASE 2: Arithmetic Operations
# ==========================================
code_addition = """
a = 10
b = 20
c = a + b
print("The addition is:", c)
print("Welcome to Langchain Tools")
"""
result_2 = run_dynamic_python(code_addition)
print(result_2)


# ==========================================
# TEST CASE 3: Standard Logging Strings
# ==========================================
code_strings = """
print("Welcome to Langchain Tools")
print("Hello World, this is Langchain Python REPL Tool")
"""
result_3 = run_dynamic_python(code_strings)
print(result_3)
```

```python
import os
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.tools import Tool
from langchain_experimental.utilities import PythonREPL
from langchain_community.tools.shell.tool import ShellTool

# FIX: Import from the explicit legacy package
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor

# 1. Setup API environment
import os
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
llm = ChatOpenAI(model="gpt-5.4-nano", temperature=0)

# 2. Build Tools
base_shell_tool = ShellTool()
shell_agent_tool = Tool(
    name="terminal_shell",
    description="Execute bash commands. Input must be a pure string.",
    func=lambda command: base_shell_tool.run({"commands": command})
)

python_repl_utility = PythonREPL()
python_repl_utility.globals = globals()
python_repl_utility.locals = locals()
python_repl = Tool(
    name="python_repl",
    description="Execute python code. Use print() to check output.",
    func=python_repl_utility.run
)

tools = [shell_agent_tool, python_repl]

# 3. Create prompt with valid placeholder object
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Use tools to run python code or terminal commands.If the tool is not available for user prompt, use your pretrained knowledge to answer user query."),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# 4. Initialize legacy structure
agent = create_tool_calling_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)

# 5. Run Agent
user_query = input("What would you like to execute? \n")
response = agent_executor.invoke({"input": user_query})

print(f"\nFinal Response:\n{response['output']}")
```

```python
import os
import sys
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.tools import Tool
from langchain_experimental.utilities import PythonREPL
from langchain_community.tools.shell.tool import ShellTool

# Import for Wikipedia and DuckDuckGo search capabilities
from langchain_community.utilities import WikipediaAPIWrapper
from langchain_community.tools import WikipediaQueryRun, DuckDuckGoSearchRun

# Import from the explicit legacy package
from langchain_classic.agents import create_tool_calling_agent, AgentExecutor

# 1. Setup API environment
import os
os.environ["OPENAI_API_KEY"] = userdata.get('OPENAI_API_KEY')
os.environ["WIKIPEDIA_USER_AGENT"] = "my-langchain-agent/1.0 (contact@example.com)"

llm = ChatOpenAI(model="gpt-5.4-nano", temperature=0)

# 2. Build Tools
base_shell_tool = ShellTool()
shell_agent_tool = Tool(
    name="terminal_shell",
    description="Execute bash commands. Input must be a pure string.",
    func=lambda command: base_shell_tool.run({"commands": command})
)

python_repl_utility = PythonREPL()
python_repl_utility.globals = globals()
python_repl_utility.locals = locals()
python_repl = Tool(
    name="python_repl",
    description="Execute python code. Use print() to check output.",
    func=python_repl_utility.run
)

wiki_api_wrapper = WikipediaAPIWrapper()
wiki_tool = WikipediaQueryRun(api_wrapper=wiki_api_wrapper)

ddg_search_tool = DuckDuckGoSearchRun()

tools = [shell_agent_tool, python_repl, wiki_tool, ddg_search_tool]

# 3. Create prompt with valid placeholder object
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant. Use tools to search the web, lookup wikipedia, run python code, or run terminal commands. If a specific tool is not available, do not use your pretrained knowledge; just say 'I can't answer as the tool is not available'."),
    ("human", "{input}"),
    MessagesPlaceholder(variable_name="agent_scratchpad"),
])

# 4. Initialize legacy structure
agent = create_tool_calling_agent(llm, tools, prompt)
agent_executor = AgentExecutor(agent=agent, tools=tools, verbose=True, handle_parsing_errors=True)

# 5. DYNAMIC CHAT LOOP
print("=" * 60)
print(" Agent Session Initialized! Type 'exit' or 'quit' to stop. ")
print("=" * 60)

while True:
    try:
        # Prompt the user for input
        user_query = input("\nYou: ")

        # Check for termination commands (case-insensitive and stripped of whitespace)
        if user_query.strip().lower() in ['exit', 'quit']:
            print("\nExiting session. Goodbye!")
            break

        # Skip empty inputs
        if not user_query.strip():
            continue

        # Execute the agent loop
        response = agent_executor.invoke({"input": user_query})
        print(f"\nAgent:\n{response['output']}")

    except (KeyboardInterrupt, EOFError):
        # Gracefully handle Ctrl+C or Ctrl+D shortcuts in terminal
        print("\n\nSession interrupted by user. Goodbye!")
        sys.exit(0)
    except Exception as e:
        print(f"\nAn error occurred: {e}")
```
