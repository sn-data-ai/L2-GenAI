```python
# 1 - Stateless Chat-Bot
```

```python
system_role = input("Enter the system role (e.g. You are a helpful assistant): \n")

while True:
    user_input = input("You/Human: \n")
    if user_input.lower() == "exit":
        break

    # Build a fresh, independent list of messages for every single turn
    stateless_messages = [
        SystemMessage(content=system_role),
        HumanMessage(content=user_input)
    ]

    # Format the prompt using only the current conversation turn
    formatted_input_prompt = chat_prompt_template.from_messages(stateless_messages)

    # Invoke the model with the stateless prompt
    response = model.invoke(formatted_input_prompt.invoke({}))

    print("\nThe AI / LLM response is : \n")
    print(response.content)

print("Exiting the Chat - Good Bye!")
```

```python
# Example - 3 - Dynamic Context aware Chat-bot using MessagePlaceHolder
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

print("Welcome to Context Aware Chat-bot using MessagePlaceHolder")

# 1. Define the System Role
system_role = input("Enter the system role (e.g. You are a helpful assistant, your responses are brief and accurate.): \n")

# 2. Initialize the chat history
chat_history = []

# 3. Setup the template with a MessagesPlaceholder to dynamically inject history
chat_template = ChatPromptTemplate.from_messages(
    [
        ("system", system_role),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{prompt}")
    ]
)

while True:
    user_input = input("\nYou/Human: \n")
    if user_input.lower() == "exit":
        break

    # 4. Invoke the template, passing both the history list and current input string
    formatted_input_prompt = chat_template.invoke({
        "chat_history": chat_history,
        "prompt": user_input
    })

    # 5. Get the LLM response
    response = model.invoke(formatted_input_prompt)

    print("\nThe AI / LLM response is : \n")
    print(response.content)

    # 6. Append both the human prompt and AI response to the message history
    chat_history.append(HumanMessage(content=user_input))
    chat_history.append(AIMessage(content=response.content))

print("Exiting the Chat - Good Bye!")
```

```python
# Example - 4 -> The above chatbot with tikton library to count the tokens used locally with each iteration.
```

```python
import tiktoken
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

# Initialize tiktoken encoder (change to your model's specific encoding if needed)
# "cl100k_base" works for gpt-3.5-turbo / gpt-4. "o200k_base" is used for gpt-4o.
encoder = tiktoken.get_encoding("cl100k_base")

def count_tokens(text: str) -> int:
    """Helper function to return the local token count of a string string."""
    return len(encoder.encode(text))

print("Welcome to Context Aware Chat-bot using MessagePlaceHolder")

# 1. Define the System Role
system_role = input("Enter the system role (e.g. You are a helpful assistant): \n")

# 2. Initialize the chat history
chat_history = []

# 3. Setup the template with a MessagesPlaceholder to dynamically inject history
chat_template = ChatPromptTemplate.from_messages(
    [
        ("system", system_role),
        MessagesPlaceholder(variable_name="chat_history"),
        ("human", "{prompt}")
    ]
)

while True:
    user_input = input("\nYou/Human: \n")
    if user_input.lower() == "exit":
        break

    # 4. Invoke the template, passing both the history list and current input string
    formatted_input_prompt = chat_template.invoke({
        "chat_history": chat_history,
        "prompt": user_input
    })

    # 5. Extract text from the formatted LangChain prompt to count input tokens locally
    # We combine the content of all messages being sent to the model this turn
    full_input_text = "\n".join([msg.content for msg in formatted_input_prompt.to_messages()])
    input_tokens = count_tokens(full_input_text)

    # 6. Get the LLM response
    response = model.invoke(formatted_input_prompt)

    # 7. Count output tokens locally from the raw string response
    output_tokens = count_tokens(response.content)

    print("\nThe AI / LLM response is : \n")
    print(response.content)

    # Display the local token usage analytics
    print(f"\n[Local Token Usage] Input Prompt: {input_tokens} tokens | Output Response: {output_tokens} tokens")

    # 8. Append both the human prompt and AI response to the message history
    chat_history.append(HumanMessage(content=user_input))
    chat_history.append(AIMessage(content=response.content))

print("Exiting the Chat - Good Bye!")
```

```python
import gradio as gr
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.messages import HumanMessage, AIMessage

# --- 1. System Role Action Logic ---
def set_role(role_text):
    if not role_text.strip():
        return gr.update(), "⚠️ Please enter a valid system role before proceeding."
    return role_text, f"✅ System Role configured successfully: '{role_text[:30]}...'"

# --- 2. Chat Processing Logic ---
def predict(system_role_state, user_input, chat_history):
    # Guard clause: Ensure system role state exists
    if not system_role_state:
        return user_input, chat_history, "❌ Error: You must click 'Set System Role' before chatting!"

    if not user_input.strip():
        return "", chat_history, ""

    # Setup LangChain Template
    chat_template = ChatPromptTemplate.from_messages(
        [
            ("system", system_role_state),
            MessagesPlaceholder(variable_name="chat_history"),
            ("human", "{prompt}")
        ]
    )

    # Convert Gradio's chat format back to LangChain objects
    langchain_history = []
    for turn in chat_history:
        # Modern Gradio handles history using OpenAI-style dicts
        if isinstance(turn, dict):
            role = turn.get("role")
            content = turn.get("content")
            if role == "user":
                langchain_history.append(HumanMessage(content=content))
            elif role == "assistant":
                langchain_history.append(AIMessage(content=content))
        # Handle cases where your local Gradio version provides object structures
        elif hasattr(turn, "role"):
            if turn.role == "user":
                langchain_history.append(HumanMessage(content=turn.content))
            elif turn.role == "assistant":
                langchain_history.append(AIMessage(content=turn.content))
        # Legacy tuple structure fallback
        elif isinstance(turn, (list, tuple)) and len(turn) == 2:
            langchain_history.append(HumanMessage(content=turn[0]))
            langchain_history.append(AIMessage(content=turn[1]))

    # Format and execute LangChain prompt components
    formatted_input_prompt = chat_template.invoke({
        "chat_history": langchain_history,
        "prompt": user_input
    })

    response = model.invoke(formatted_input_prompt)

    # Append the newest exchange into Gradio's UI component history
    chat_history.append({"role": "user", "content": user_input})
    chat_history.append({"role": "assistant", "content": response.content})

    return "", chat_history, ""

# --- 3. UI Structural Component Definitions ---
with gr.Blocks(title="Context-Aware Chatbot") as demo:
    gr.Markdown("## 🤖 Context Aware Chat-bot using MessagePlaceHolder")

    system_role_state = gr.State(value="")

    # SECTION 1: Role Configuration Setup
    with gr.Group():
        gr.Markdown("### Step 1: Assign System Persona")
        with gr.Row():
            system_role_input = gr.Textbox(
                label="System Role Definition",
                value="You are a helpful assistant, your responses are brief and accurate.",
                lines=3,
                placeholder="Instruct the AI behavior..."
            )
        with gr.Row():
            set_role_btn = gr.Button("🔒 Set System Role", variant="primary")
            status_output = gr.Markdown("*Status: Waiting for role assignment...*")

    gr.Markdown("---")

    # SECTION 2: Chat Interaction Area
    with gr.Group():
        gr.Markdown("### Step 2: Prompt Submission")
        with gr.Row():
            user_message_input = gr.Textbox(
                label="You / Human Input",
                lines=3,
                placeholder="Type your message here..."
            )

        # FIXED: Removed 'type' argument to maintain native compatibility across Gradio builds
        chatbot_display = gr.Chatbot(
            label="AI / LLM Interaction History Log",
            height=450
        )

        submit_chat_btn = gr.Button("🚀 Submit Prompt")

    # --- 4. Event Binding Configuration Actions ---
    set_role_btn.click(
        fn=set_role,
        inputs=[system_role_input],
        outputs=[system_role_state, status_output]
    )

    submit_chat_btn.click(
        fn=predict,
        inputs=[system_role_state, user_message_input, chatbot_display],
        outputs=[user_message_input, chatbot_display, status_output]
    )
    user_message_input.submit(
        fn=predict,
        inputs=[system_role_state, user_message_input, chatbot_display],
        outputs=[user_message_input, chatbot_display, status_output]
    )

if __name__ == "__main__":
    demo.launch()
```
