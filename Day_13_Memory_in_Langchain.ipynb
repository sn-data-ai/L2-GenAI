```python
# Example 3 - Context aware chat bot - Dynamic & using Conversation Buffer Memory - entire conversation storage
```

```python
# Define a conversation chain

conversation = ConversationChain(llm=model, memory=ConversationBufferMemory(), verbose=False)
print("\nWelcome to Context Aware Chat-bot! Type 'exit' or 'quit' to stop the chatbot!\n")

# Dynamic conversation loop

while True:
  user_input = input("Human: ")

  if user_input.lower() in ["exit", "quit", "q"]:
    print("Good Bye!")
    break

  response = conversation.predict(input=user_input)
  print(f"AI: {response}")
```

```python
# ConversationSummaryMemory
```

```python
# Define a conversation chain

conversation_summary = ConversationChain(llm=model,memory=ConversationSummaryMemory(llm=model))
print("\nWelcome to Context Aware Chat-bot! Type 'exit' or 'quit' to stop the chatbot!\n")

# Dynamic conversation loop

while True:
  user_input = input("Human: ")

  if user_input.lower() in ["exit", "quit", "q"]:
    print("Good Bye!")
    break

  # Display the Models response to user prompt / user question
  response = conversation_summary.predict(input=user_input)
  print(f"AI: {response}")

  # Display running summary of the conversation
  print(f"Running Summary: {conversation_summary.memory.buffer}")
```

```python
# Example 6 - The baove chat bot to store conversation on disk / local storage
```

```python
import os
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationSummaryMemory
from langchain_classic.memory.chat_message_histories import FileChatMessageHistory

# 1. Define a secure path in your Google Colab environment
# This stores your conversation log locally as a JSON file
colab_storage_path = "/content/chat_history.json"

# Ensure the directory exists (helpful if using custom subfolders)
os.makedirs(os.path.dirname(colab_storage_path), exist_ok=True)

# 2. Initialize the file-backed message history tracker
file_history = FileChatMessageHistory(file_path=colab_storage_path)

# 3. Inject the file tracker into your Summary Memory framework
conversation_summary = ConversationChain(
    llm=model,
    memory=ConversationSummaryMemory(
        llm=model,
        chat_memory=file_history  # Binds file storage to memory
    )
)

print("\nWelcome to Context Aware Chat-bot! Type 'exit' or 'quit' to stop the chatbot!\n")

# 4. Dynamic conversation loop
while True:
  user_input = input("Human: ")

  if user_input.lower() in ["exit", "quit", "q"]:
    print("Good Bye!")
    break

  # Generate and print the AI response
  response = conversation_summary.predict(input=user_input)
  print(f"AI: {response}")

  # Display the background summary of the conversation
  print(f"Running Summary: {conversation_summary.memory.buffer}\n")
```
