**Models in Langchain**

```python
# Plug in various LLM models from different families easily -
# without changing the complete code

# To achieve this - we use 'Models' in langchain. We can plug in any LLM model from any family and use it in the same code without changing the complete code.

# pip install langchain, langchain_ollama, langchain_openai

# Models in langchain - help unify coding application - calling LLMs
# functions to interact with different LLMs in a consistent way
# 1 invoke() -> calls the LLM from any family - with the given input and returns the output
```

```python
# Example 1 (without Models in Langchain) - Ollama Cloud Model - gpt-oss:20b-cloud
```

```python
!pip install ollama -q
```

```python
# Define prompt

prompt = "What is the capital of USA?"
```

```python
prompt
```

```python
# Imports
from google.colab import userdata
from ollama import Client

# Access AOI Key and Initialize the client object

ollama_api_key = userdata.get("OLLAMA_API_KEY")

client = Client(
    host="https://ollama.com",
    headers = {"Authorization": f"Bearer {ollama_api_key}"}
)
print("Ollama Client Initialized")
```

```python
# Ollama chat bot below
```

```python
response_ollama = client.chat(
        model = 'gpt-oss:20b:cloud',
        messages = [
        {"role":"user", "content": prompt}
        ]
    )

print("The response from Ollama cloud llm is: \n")
print(response_ollama['message']['content'])
```

```python
# OpenAI chat bot below
```

```python
from openai import OpenAI
from google.colab import userdata

openai_api_key = userdata.get("OPENAI_API_KEY")

client = OpenAI(api_key=openai_api_key)
print("OpenAI Client Initialized")
```

```python
response_openai = client.responses.create(
    model="gpt-5.4-nano",
    input=prompt
)
print("The response from OpenAI cloud llm is: \n")
print(response_openai.output_text)
```

```python
print("Example 1 Output / Response: \n")
print("The response from Ollama cloud llm is: \n")
print(response_ollama['message']['content'])
print("Example 2 Output / Response: \n")
print("The response from OpenAI cloud llm is: \n")
print(response_openai.output_text)
```

```python
# Example 3 - Using Langchain - Models - invoke()
# Interact with Ollama/OpenAI cloud LLMs via Langchain
```

```python
!pip install langchain langchain-openai langchain-ollama -q
```

```python
from google.colab import userdata
from langchain_ollama import ChatOllama

# Setup API Key access
ollama_api_key = userdata.get("OLLAMA_API_KEY")

# Initialize the langchain ChatOllama Object

client_ollama = ChatOllama(
    model = 'gpt-oss:20b:cloud',
    base_url="https://ollama.com",
    client_kwargs = {
    "headers" : {
        "Authorization": f"Bearer {ollama_api_key}"}
    }
)
print("Langchain ChatOllama Initialized")
```

```python
response_ollama = client_ollama.invoke(prompt)
print("The response from Ollama cloud llm is: \n")
print(response_ollama.content)
```

```python
from google.colab import userdata
from langchain_openai import ChatOpenAI

# Setup API Key access
openai_api_key = userdata.get("OPENAI_API_KEY")

client_openai = ChatOpenAI(
    model = 'gpt-5.4-nano',
    api_key = openai_api_key
)
print("Langchain ChatOpenAI Initialized")
```

```python
response_openai = client_openai.invoke(prompt)
print("The response from OpenAI cloud llm is: \n")
print(response_openai.content)
```

```python
# Example 3 Output
response_ollama = client.invoke("Whats the capital of India?")
print("The response from Ollama cloud llm is: \n")
print(response_ollama.content)
```

```python
# Example 4 Output
response_openai = client.invoke("Whats the capital of India?")
print("The response from Openai cloud llm is: \n")
print(response_openai.content)
```

```python
user_prompt = input("Enter your prompt here: \n")

response_ollama = client_ollama.invoke(user_prompt)
print("The response from Ollama cloud llm is: \n")
print(response_ollama.content)

response_openai = client_openai.invoke(user_prompt)
print("The response from Openai cloud llm is: \n")
print(response_openai.content)
```
